```
import os
import IPython
from IPython.display import Audio
import librosa
import numpy as np
from pathlib import Path
from resemblyzer import VoiceEncoder, preprocess_wav

from pydub import effects  
from pydub import AudioSegment

def cosine_similarity(x, y):
    x = np.array(x).reshape(-1)
    y = np.array(y).reshape(-1)
    return np.dot(x, y) / (np.sqrt(np.dot(x, x)) * np.sqrt(np.dot(y, y)))

encoder = VoiceEncoder()

def compute_emb(path, normalize=True, target_dbfs=-27): #target_dbfs=-24.632442475923607
    if normalize:
      song = AudioSegment.from_file(path)
      change_in_dBFS = target_dbfs - song.dBFS
      normalized_sound = song.apply_gain(change_in_dBFS)
      normalized_sound.export(path, format=path[-3:])

    fpath = Path(path)
    wav = preprocess_wav(fpath)
    embed = encoder.embed_utterance(wav)
    # print(embed.shape)
    return embed
```

```
import os
import librosa

import IPython
import numpy as np
from IPython.display import Audio

def extract_features(ref_file):
  y, sr = librosa.load(ref_file, sr=ap.sample_rate)
  spec = ap.spectrogram(y)
  spec = torch.FloatTensor(spec).unsqueeze(0)
  return y, spec
```

```
import scipy.stats as st
import random
import numpy as np
import random
import pandas as pd

def set_seed(seed=42):
  np.random.seed(seed)
  random.seed(seed)
  torch.random.manual_seed(seed)
  torch.cuda.manual_seed(seed)
  torch.manual_seed(seed)
  # torch.set_deterministic(True)
  torch.backends.cudnn.deterministic = True
  torch.backends.cudnn.benchmark = False

def show_result(similarity):
  print("Results:")
  all_secs = []         
  for key in similarity.keys():
    speaker_secs = np.array(similarity[key]).mean()
    low_b, high_b = st.t.interval(0.95, len(similarity[key])-1, loc=speaker_secs, scale=st.sem(similarity[key]))
    confience_interval_more_less = round(speaker_secs-low_b, 4)
    print("Speaker:", key, "SECS:", speaker_secs, "+-", confience_interval_more_less)
    all_secs += similarity[key]

  secs_all = np.array(all_secs).mean()
  low_b, high_b = st.t.interval(0.95, len(all_secs)-1, loc=secs_all, scale=st.sem(all_secs))
  confience_interval_more_less = round(secs_all-low_b, 4)
  print("All SECS:", secs_all, "+-", confience_interval_more_less)

def compute_SECS_and_save_audios(speaker_references, speaker_list1, speaker_list2, OUTPUT_WAV_PATH, num_samples_per_spk_to_show=1): 
  # set seed for reproducibility
  set_seed()

  # create paths 
  os.makedirs(os.path.join(OUTPUT_WAV_PATH, "generated"), exist_ok=True)
  os.makedirs(os.path.join(OUTPUT_WAV_PATH, "reference"), exist_ok=True)
  # os.makedirs(os.path.join(OUTPUT_WAV_PATH, "../", "GT"), exist_ok=True)

  similarity = {}
  metadata = []
  for spk1 in speaker_list1:
    num_samples_spk = 0 
    for spk2 in speaker_list2:
      # ignore some speaker
      if spk1 == spk2:
        continue
      for _ in range(refs_per_speaker):
        ref_wav_path = random.choice(speaker_references[spk1])
        target_speaker_wav_path = random.choice(speaker_references[spk2])

        ref_wav, ref_sr = librosa.load(ref_wav_path, sr=ap.sample_rate)
        t_ref_wav, t_ref_sr = librosa.load(target_speaker_wav_path, sr=ap.sample_rate)
        
        ref_speaker_name = os.path.basename(ref_wav_path).split("_")[0]
        target_speaker_name = os.path.basename(target_speaker_wav_path).split("_")[0]
        print("Voice conversion with Speaker: ", target_speaker_name, " as reference speaker", ref_speaker_name)

        # get precompute embedding
        d_vector_ref = model.speaker_manager.get_d_vector_by_clip(os.path.basename(ref_wav_path))
        d_vector_target = model.speaker_manager.get_d_vector_by_clip(os.path.basename(target_speaker_wav_path))

        d_vector_ref = torch.FloatTensor(d_vector_ref).unsqueeze(0)
        d_vector_target = torch.FloatTensor(d_vector_target).unsqueeze(0)

        _, ref_spectrogram = extract_features(ref_wav_path)

        y_lengths = torch.tensor([ref_spectrogram.size(-1)]).cuda()


        # run the model
        wav, _, _ = model.voice_conversion(ref_spectrogram.cuda(), y_lengths, d_vector_ref.cuda(), d_vector_target.cuda())
        wav = wav.squeeze().cpu().detach().numpy()
        # show audios
        if num_samples_spk < num_samples_per_spk_to_show:
          print("Audio ref:")
          IPython.display.display(Audio(ref_wav, rate=ref_sr))
          print("Audio from target Speaker:")
          IPython.display.display(Audio(t_ref_wav, rate=t_ref_sr))
          print("Converted Audio: ")
          IPython.display.display(Audio(wav, rate=ap.sample_rate))
          num_samples_spk += 1

        # save the results
        out_path = os.path.join(OUTPUT_WAV_PATH, "generated", ref_speaker_name+'_to_'+target_speaker_name+'_sample_'+os.path.basename(target_speaker_wav_path))
        ap.save_wav(wav, out_path)
        out_path_target_speaker_ref = os.path.join(OUTPUT_WAV_PATH, "reference", os.path.basename(target_speaker_wav_path))
        ap.save_wav(t_ref_wav, out_path_target_speaker_ref)

        # out_path_ref_used = os.path.join(OUTPUT_WAV_PATH, "../", "GT", os.path.basename(target_speaker_wav_path))
        # ap.save_wav(ref_wav, out_path_ref_used)
        
        
        # compute embedding
        ref_spk_emb = compute_emb(out_path_target_speaker_ref)
        gen_emb = compute_emb(out_path)
        # compute SECS
        secs = cosine_similarity(ref_spk_emb, gen_emb)
        print("SECS:", secs)
        if target_speaker_name in similarity.keys():
          similarity[target_speaker_name] += [secs]
        else:
          similarity[target_speaker_name] = [secs]
          
        metadata.append([out_path_target_speaker_ref, out_path])


  show_result(similarity)

  df = pd.DataFrame(metadata, columns=["target_spk_ref", "generated"])
  df.to_csv(os.path.join(OUTPUT_WAV_PATH, "metadata.csv"), index=False)
  return metadata
```

# **Instance TTS Model and utils**

```
def load_model(model_path):
  global model
  cp = torch.load(model_path, map_location=torch.device('cpu'))
  # remove speaker encoder
  model_weights = cp['model'].copy()
  for key in list(model_weights.keys()):
    if "speaker_encoder" in key:
      del model_weights[key]

  model.load_state_dict(model_weights)
  model.eval()

  if USE_CUDA:
      model = model.cuda()

  model.inference_noise_scale = 0.333 # defines the noise variance applied to the random z vector at inference.
  model.inference_noise_scale_dp = 0.333 # defines the noise variance applied to the duration predictor z vector at inference.
```

```
# load the config
C = load_config(CONFIG_PATH)

# load the audio processor
ap = AudioProcessor(**C.audio)

speaker_embedding = None

# synthesize voice
use_griffin_lim = False

C.model_args['d_vector_file'] = TTS_SPEAKERS
C.model_args['use_speaker_encoder_as_loss'] = False

model = setup_model(C)
model.language_manager.set_language_ids_from_file(TTS_LANGUAGES)


```

```
import glob, os
audio_min_len = 3 # seconds
speakers_references = {}
Dataset_path = "Test-Dataset/"

for f in glob.glob(os.path.join(Dataset_path, "**/*.wav"), recursive=True):
    y, sr = librosa.load(f, sr=16000)
    audio_len = librosa.get_duration(y=y, sr=sr)
     
    if audio_len < audio_min_len:
      continue
    speaker_name = os.path.basename(f).split("_")[0]
    if speaker_name not in speakers_references:
      speakers_references[speaker_name] = [f]
    else:
        speakers_references[speaker_name].append(f)
print(speakers_references)
```
